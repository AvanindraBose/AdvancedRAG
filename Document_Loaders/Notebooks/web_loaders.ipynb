# Recursive Loader

-> A loader which is used to load the child links of the base url. 

-> It should be used with lazy load as we may have a lot of child links.

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

loader = RecursiveUrlLoader(url=base_url,
                            max_depth=2)
```

```python
document = loader.lazy_load()
```

```python
type(document)
```

```python
counter = 0


for doc in document :
    for counter in range(10):
        print(doc.page_content[:500])
        print(doc.metadata)
        print("\n\n")
```
