```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
```

```python
file_path = Path("../Knowledge_Source/apparels.json")
file_path.exists()
```

```python
jq = '''
.products[] | . + {
  text: (
    "Product: " + .productName +
    ". Description: " + .Description
  )
}
'''
```

### Creating a Custom Markdown Function to display only necessary Stuff

```python
def metadata_content(knowledge_source : dict , metadata : dict) -> dict :
    metadata['price'] = knowledge_source['price']
    metadata['category'] = knowledge_source['category']
    metadata['brand'] = knowledge_source['manufacturer']

    return metadata
```

```python
loader = JSONLoader(
    file_path=file_path,
    jq_schema=jq,
    content_key="text",
    metadata_func=metadata_content
)
```

```python
document = loader.load()
```
