# PyPDFLoader

```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
```

```python
file_path = Path("../Knowledge_Source/attention_is_all_you_need.pdf")
file_path.exists()
```

```python
loader = PyPDFLoader(file_path=file_path,
                     mode='page')
```

```python
document = loader.load()
```

```python
print(type(document))

print(type(document[0]))
```

```python
print(document[3])
```

```python
print(document[3].page_content)
```

```python
print(document[3].metadata)
```

# Observation

1. PyPDF Loader clearly struggels with images. We have an option of using Image OCR we will see that next.

## Using Image Capabilities of PyPDF Loader Using Image OCR

```python
from langchain_community.document_loaders.parsers import TesseractBlobParser,RapidOCRBlobParser
```

```python
image_loader = PyPDFLoader(
    file_path=file_path,
    mode='page',
    images_parser = RapidOCRBlobParser(),
    images_inner_format='html-img'
)
```

```python
image_doc = image_loader.load()
```

```python
print(image_doc[3].page_content)
```

```python
print(image_doc[2].page_content)
```

# Observation

1. With RapidOCR image Parser PyPDFLoader was able to extract the image in html format at the end of the Page. But in cases where Images or tables are present we should usually prefer PyMUPDFLoader , PDFMiner , Docling(__New thing__)
