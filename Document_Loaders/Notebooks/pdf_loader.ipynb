# PyPDFLoader

```python
from langchain_community.document_loaders.pdf import PyPDFLoader
from pathlib import Path
```

```python
file_path = Path("../Knowledge_Source/attention_is_all_you_need.pdf")
file_path.exists()
```

```python
loader = PyPDFLoader(file_path=file_path,
                     mode='page')
```

```python
document = loader.load()
```

```python
print(type(document))

print(type(document[0]))
```

```python
print(document[3])
```

```python
print(document[3].page_content)
```

```python
print(document[3].metadata)
```

# Observation

1. PyPDF Loader clearly struggels with images. We have an option of using Image OCR we will see that next.

## Using Image Capabilities of PyPDF Loader Using Image OCR

```python
from langchain_community.document_loaders.parsers import TesseractBlobParser,RapidOCRBlobParser
```

```python
image_loader = PyPDFLoader(
    file_path=file_path,
    mode='page',
    images_parser = RapidOCRBlobParser(),
    images_inner_format='html-img'
)
```

```python
image_doc = image_loader.load()
```

```python
print(image_doc[3].page_content)
```

```python
print(image_doc[2].page_content)
```

# Observation

1. With RapidOCR image Parser PyPDFLoader was able to extract the image in html format at the end of the Page. But in cases where Images or tables are present we should usually prefer PyMUPDFLoader , PDFMiner , Docling(__New thing__) 

# PDF Miner

```python
from langchain_community.document_loaders.pdf import PDFMinerLoader
```

```python
miner = PDFMinerLoader(
    file_path=file_path,
    mode='page',
    images_parser=RapidOCRBlobParser(),
    images_inner_format='html-img'
)
```

```python
doc_image = miner.load()
```

```python
print(doc_image[0])
```

```python
print(doc_image[2].page_content)
```

```python
print(doc_image[2].metadata)
```

```python
print(doc_image[3].page_content)
```

```python
print(doc_image[5].page_content)
```

```python
miner_md = PDFMinerLoader(
    file_path=file_path,
    mode='page',
    images_parser=RapidOCRBlobParser(),
    images_inner_format='markdown-img'
)
```

```python
doc_img1 = miner_md.load()
```

```python
print(doc_img1[3].page_content)
```

# Observations

1. PDF Miner is performing comparitively with PyPDF for Images. But it works better for Images. 
2. It was able to extract the information correctly out of the table.
3. It struggeled a bit with correct text extraction from Image, but it is comparatively better.
