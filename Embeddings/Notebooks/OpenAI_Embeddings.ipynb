```python
loader = PyPDFLoader(
    file_path=file_path
)
```

## Embedding Query

```python
query = "Time will change. And I show myself what can I achieve."
```

```python
embedder_query = OpenAIEmbeddings(
    model='text-embedding-3-small'
)
```

```python
embedding = embedder_query.embed_query(text=query)
```
