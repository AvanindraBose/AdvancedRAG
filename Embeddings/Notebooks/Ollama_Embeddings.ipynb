# Langchain - Ollama

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
from langchain_ollama.embeddings import OllamaEmbeddings
from pathlib import Path
```

```python
file_path = Path("../Documents/Openclaw_Research_Report.pdf")
file_path.exists()
```

```python
loader = PyPDFLoader(
    file_path = file_path
)
```

```python
documents = loader.load()
```

```python
chunker = RecursiveCharacterTextSplitter(
    chunk_size = 300,
    chunk_overlap = 20
)
```

```python
chunks = chunker.split_documents(documents)
```

```python
len(chunks)
```

```python
# embed documents

text_documents = [doc.page_content for doc in chunks]

print(len(text_documents))
```

```python
# create the embedder

langchain_embedder = OllamaEmbeddings(
    model="embeddinggemma"
)
```

```python
langchain_document_embeddings = langchain_embedder.embed_documents(
    texts=text_documents
)
```

```python
len(langchain_document_embeddings)
```
